```python
import numpy as np

def tridiag_toeplitz(n, a, b=-1.0):
    A = np.zeros((n, n))
    i = np.arange(n)
    A[i, i] = a
    A[i[1:], i[:-1]] = b
    A[i[:-1], i[1:]] = b
    return A

def kappa2(A):
    # For symmetric A, 2-norm condition number is ratio of extreme eigenvalues
    w = np.linalg.eigvalsh(A)
    return float(max(abs(w)) / min(abs(w)))

n = 6
for a in [1.82, 1.85, 2.0, 2.5, 3.0, 5.0, 10.0]:
    A = tridiag_toeplitz(n, a, b=-1.0)
    print(a, kappa2(A))
    print(A)
```

```python
import numpy as np

def tridiag_toeplitz(n, a, b=-1.0):
    A = np.zeros((n, n))
    i = np.arange(n)
    A[i, i] = a
    A[i[1:], i[:-1]] = b
    A[i[:-1], i[1:]] = b
    return A

def neumann_laplacian(n):
    A = np.zeros((n, n))
    A[0,0] = 1.0
    A[-1,-1] = 1.0
    A[np.arange(1,n-1), np.arange(1,n-1)] = 2.0
    i = np.arange(n-1)
    A[i, i+1] = -1.0
    A[i+1, i] = -1.0
    return A

def cond2_spd(A):
    w = np.linalg.eigvalsh(A)
    print("eigen:", w)
    return float(w.max() / w.min())

n = 6
c = 2*np.cos(np.pi/(n+1))
for delta in [100.0, 1.0, 0.5, 1e-1, 1e-2, 1e-4, 1e-6, 1e-8]:
    A = tridiag_toeplitz(n, a=c+delta, b=-1.0)
    print(f"delta={delta:>8}, kappa2={cond2_spd(A):.3e}")
    print(A)

#for eps in [1e-2, 1e-4, 1e-6]:
#    A = neumann_laplacian(n) + eps*np.eye(n)
#    print(f"eps  ={eps:>8}, kappa2={cond2_spd(A):.3e}")
```
